```python
# Mapear variações para o valor correto e susbstitui pelo valores certos
mapa = {
    'corporativo':                     'Corporativo',
    'pessoa fisica':                   'Pessoa Física',
    'PF':                              'Pessoa Física',
    'pf' :                              'Pessoa Física'

}
df_clientes['tipo_cliente'] = df_clientes['tipo_cliente'].map(mapa).fillna(df_clientes['tipo_cliente'])#joga as informações do dicionario mapa no df. filne mantem os valores corretos que não precisam de substituição

print('Após mapeamento:')
print(sorted(df_clientes['tipo_cliente'].unique()))
```

```python
df_clientes.loc[
    df_clientes.duplicated(subset=['nome'], keep=False),
    'nome'
]

```

```python
# nome: verificar espaços invisíveis
print('\n--- nomes com espaço nas bordas ---')
mascara = df_clientes['nome'].str.startswith(' ') | df_clientes['nome'].str.endswith(' ')
(df_clientes.loc[mascara, 'nome'].tolist())
```

```python
df_clientes['nome'] = df_clientes['nome'].str.strip().str.title()
print(sorted(df_clientes['nome'].unique()))
```
